# Exploratory Data Analysis

### gender 

```python
df.columns
```

```python
# plotting a bar chart for Gender
plt.figure(figsize=(5,3)) 
ax = sns.countplot(x = "Gender", data = df, width = 0.7) 
```

```python
# plotting a bar chart for Gender and it's count
ax = sns.countplot(x = "Gender", data = df)

for bar in ax.containers:
    ax.bar_label(bar)
```

```python
sails_data = df.groupby(['Gender'],as_index = False)['Amount'].sum().sort_values(by='Amount',ascending = False)
```

```python
plt.figure(figsize=(5,4))
ay = sns.barplot(x = 'Gender', y = 'Amount', data = sails_data)
```

#### From above graphs we can see that most of the buyers are females and even the purchasing power of females are greater than men

## Age 

```python
df.columns
```

```python
ax = sns.countplot(x = 'Age Group',hue= "Gender" ,data= df) 
# hue = "Gender" isse apan ko ek right side me ek box dikh rha hai female or male ka vo show krne ke kaam aata hai.

for bar in ax.containers:
    ax.bar_label(bar)
```

```python
# it define the how much shopping is done by different age group and who done the highest shopping b/w male or female .
```

```python
# Total Amount v/s Age Group
sails_age = df.groupby(["Age Group"], as_index = False)["Amount"].sum().sort_values(by = "Amount" ,ascending = False)

sns.barplot(x = 'Age Group',y= 'Amount', hue = "Age Group" ,data = sails_age)
```

#### From above graphs we can see that most of the buyers are of age group between 26-35 yrs female

## State

```python
# total number of order from top 10 States 
sales_state = df.groupby("State", as_index = False)["Orders"].sum().sort_values(by = "Orders", ascending = False).head(10)
plt.figure(figsize=(17,5))
sns.barplot(x = 'State',y= 'Orders' ,data = sales_state, width = 0.6, hue = "State")
```

#### it shows that which state order the highest . 

```python
# total amount/sales from top 10 states

sales_state = df.groupby(["State"], as_index = False)["Amount"].sum().sort_values(by = "Amount", ascending = False).head(10)
plt.figure(figsize=(17,5))
sns.barplot(x = "State", y = "Amount",hue = "State" ,data = sales_state)
```

#### it shows that from which state the amount is highest .

### Marital_Status

```python
plt.figure(figsize = (4,5))

ax = sns.countplot(data = df, x = "Marital_Status")

for bar in ax.containers:
    ax.bar_label(bar)
```

##### o represent's married and 1 represent single one.

#### This graph shows that married people have done shoping more than single one . 

```python
df.columns
```

```python
# Now we will check whether married people or single people have made more purchases, and we will also see whether males or females spent more

sales_data = df.groupby(["Marital_Status","Gender"], as_index = False)["Orders"].sum().sort_values(by = "Orders", ascending = False)

ax = sns.barplot(data = sales_data , x = "Marital_Status", y = "Orders", hue = "Gender")

for bar in ax.containers:
    ax.bar_label(bar)
```

```python
# ab isme apan dekhenge ki kisne jada ammount kharach kiya hai male or female me se

sales_data = df.groupby(["Marital_Status","Gender"], as_index = False) ["Amount"].sum().sort_values(by = "Amount", ascending = False)

ax = sns.barplot(data = sales_data, x = "Marital_Status", y = "Amount", hue = "Gender")

# noe we will check that kisne kitna  amount kharach kiya hai 
for bar in ax.containers:
    ax.bar_label(bar)
```

# occupation

```python
plt.figure(figsize=(21,6))

ax = sns.countplot(data = df , x = "Occupation", hue = "Occupation")

for bar in ax.containers:
    ax.bar_label(bar)
```

```python
# ab apan dekhenge ki konsi occupation ne kitna jada pesa kharach kiya hai . 
plt.figure(figsize = (21,6) )
sales_data = df.groupby(["Occupation"], as_index = False) ["Amount"].sum().sort_values(by = "Amount", ascending = False )

sns.barplot(data = sales_data , x = "Occupation", y = "Amount", hue = "Occupation", width = 0.8)
```

#### From above graphs we can see that most of the buyers are working in IT, Healthcare and Aviation sector

## Product_Category

```python
plt.figure(figsize = (23,6))
ax = sns.countplot(data = df, x = "Product_Category", hue = "Product_Category" )

for bar in ax.containers:
    ax.bar_label(bar)
```

#### isse apan ko ye pata chal rha hai ki kis product catogery se apan ko order jada mil rha hai

```python
plt.figure(figsize = (20,7))

sales_data = df.groupby(["Product_Category"], as_index = False)["Amount"].sum().sort_values(by = "Amount", ascending = False).head(10)

sns.barplot(data = sales_data, x= "Product_Category", y = "Amount", hue = "Product_Category", width = 0.7)

```

#### isse apan ko ye pata chal rha hai ki apan ko kiss product catogery se pesa jada aa raha hai 

#### in dono graph ka conclusion ye hai ki 1st graph se apn ye dekh pa rhe h ki order clothing category se jada ho rha hai but 2nd graph se ye dikh rha hai ki jo jata ammount hai i.e jada profit apan ko food category se mil rha hai .


```python
plt.figure(figsize=(20,6))

sales_data = df.groupby(["Product_ID"], as_index = False)["Orders"].sum().sort_values(by = "Orders", ascending = False).head(10)

sns.barplot(data = sales_data, x= "Product_ID", y = "Orders", hue = "Product_ID", width = .7)
```

#### sbse jada order in product id se aaye hai 

```python
# top 10 most sold products (same thing as above)


fig1, ax1 = plt.subplots(figsize=(10,7))
top_products = df.groupby('Product_ID')['Orders'].sum().nlargest(10).sort_values(ascending=False)

# har product ke liye ek random color
colors = plt.cm.tab10.colors   # tab10 palette (10 alag-alag rang)

top_products.plot(kind='bar', color=colors, ax=ax1)

plt.show()
```
